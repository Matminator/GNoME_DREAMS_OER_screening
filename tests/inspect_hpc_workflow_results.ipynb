# Inspect HPC workflow results

Load the EXPLOG pickle written at the end of `hpc_workflow_test.py` and
inspect results visually.

Run from the project root directory.

```python
import pickle
import pandas as pd
from pathlib import Path

# Change MODE to match the run you want to inspect
MODE = "slurm_test"

pkl_path = Path("tests/TEMP_{}_hpc_workflow/explog.pkl".format(MODE))
with open(pkl_path, "rb") as f:
    EXPLOG = pickle.load(f)
print(f"Loaded EXPLOG from {pkl_path}")
```

```python
# Process log
display(EXPLOG.relational_frame.processes.df)
```

```python
# Candidate-level metrics
display(EXPLOG.relational_frame.candidates.df)
```

```python
# Adsorption site energies for each surface study
study = EXPLOG.relational_frame.candidates.df["study_obj"].iloc[0]
surface_studies = study.get_surface_studies()
for term_idx, ss in surface_studies.items():
    print(f"\n── Termination {term_idx} ──")
    display(ss.get_adsorption_sites_df())
```

```python
# Visualize relaxed surface
from ase.visualize import view

ss = list(surface_studies.values())[0]
atoms = ss.get_relaxed_surface()
view(atoms)
```

```python
# Visualize adsorption structures (O and OH) for each site
from ase.visualize import view

ad_dict = ss.get_adsorption_site_studies_dict()
for site_idx, ad in ad_dict.items():
    if ad.relaxed_O_atoms is not None:
        print(f"Site {site_idx} — relaxed O structure:")
        view(ad.relaxed_O_atoms)
    if ad.relaxed_OH_atoms is not None:
        print(f"Site {site_idx} — relaxed OH structure:")
        view(ad.relaxed_OH_atoms)
```
